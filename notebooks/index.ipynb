# Hardware-Oriented Math for DoA

This repository presents hardware-oriented mathematical transformations used in FPGA implementations of Direction-of-Arrival (DoA) estimation algorithms.

The notes focus on the transition from mathematical formulation to hardware-friendly computation, including numerical validation and block-level architecture concepts.
